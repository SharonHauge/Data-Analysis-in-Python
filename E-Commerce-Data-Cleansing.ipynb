## Introduction

The global sales data was originally collected by an E-Commerce Company between 1st, Dec.2010 and 9th, Dec. 2011. 
The company mainly sells gifts covering various scenarios, including but not limited to birthday gifts, wedding souvenirs, Christmas gifts, and so on. The company's customer base mainly consists of wholesalers and individual consumers, among whom wholesalers account for a considerable proportion.

Columns in the table：
- `InvoiceNo`: Invoice number. 6 digits, serving as a unique identifier for the transaction. If this code starts with the letter 'c', it indicates that the transaction has been cancelled.
- `StockCode`: Product code. A five-digit number serves as the unique identifier for the product.
- `Description`: product name.
- `Quantity`: The quantity of the product in the transaction.
- `InvoiceDate`: Invoice date and time. The date and time when the transaction occurred.
- `UnitPrice`: Unit price. The price is stated in pounds (£).
- `CustomerID`: Customer number. A 5-digit number that serves as the unique identifier for the customer.
- `Country`: Country name. The name of the country where the customer resides.

## Clean up data

Based on the results of the data evaluation section above, the data cleaning tasks that need to be performed include:

- Convert the data type of the `CustomerID` variable to string.
- Convert the data type of the `InvoiceDate` variable to datetime.
- Delete any missing observations for the `Description` variable
- Delete all duplicate data for all variables
- Replace the values of the `Country` variable, namely `'U.K.'` and `'UK'`, with `'United Kingdom'`
- Replace the value of the `Country` variable from `'USA'` to `'United States'`
- Delete observations where `Quantity` is less than or equal to 0.
- Delete data where the `UnitPrice` value is less than or equal to 0.

To distinguish between the cleaned data and the original data, we create a new variable called `cleaned_df`. All subsequent cleaning steps will be applied to `cleaned_df`.

```python
cleaned_df = df
cleaned_df
```

Convert the data type of the `CustomerID` variable to string:

```python
cleaned_df['CustomerID'] = cleaned_df['CustomerID'].astype(str)
cleaned_df['CustomerID']
```

Remove “.0” from the end of `CustomerID`:

```python
cleaned_df['CustomerID'] = cleaned_df['CustomerID'].str.slice(0, -2)
cleaned_df['CustomerID']
```

Convert the data type of the `InvoiceDate` variable to datetime type:

```python
cleaned_df['InvoiceDate'] = pd.to_datetime(cleaned_df['InvoiceDate'])
cleaned_df.InvoiceDate
```

Delete observations with missing values for the `Description` variable:

```python
cleaned_df.dropna(subset=['Description'], inplace=True)
cleaned_df.Description.isnull().sum()
```

Delete all duplicate data across all variables：

```python
cleaned_df.drop_duplicates(inplace=True)
cleaned_df[cleaned_df.duplicated()]
```

Replace the values of the `Country` variable, namely `'U.K.'` and `'UK'`, with `'United Kingdom'`.
Replace the value of the `Country` variable from `'USA'` to `'United States'`:

```python
cleaned_df.replace(['U.K.', 'UK'], 'United Kingdom', inplace=True)
cleaned_df.replace('USA', 'United States', inplace=True)
print(len(cleaned_df[cleaned_df['Country'] == 'U.K.']))
print(len(cleaned_df[cleaned_df['Country'] == 'UK']))
print(len(cleaned_df[cleaned_df['Country'] == 'USA']))
```

Delete observations where `Quantity` is less than or equal to 0.

```python
cleaned_df.drop(cleaned_df[cleaned_df.Quantity <= 0].index, inplace=True)
len(cleaned_df[cleaned_df.Quantity <= 0])
```

Delete data where the `UnitPrice` value is less than or equal to 0:

```python
cleaned_df.drop(cleaned_df[cleaned_df.UnitPrice <= 0].index, inplace=True)
len(cleaned_df[cleaned_df.UnitPrice <= 0])
```

It was found that the unit price included 0.001 pounds, which is unreasonable and likely incorrect data.

```python
cleaned_df.drop(cleaned_df[cleaned_df.UnitPrice == 0.001].index, inplace=True)
cleaned_df.UnitPrice.sort_values()
```
